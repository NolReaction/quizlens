# Метаданные набора

Дату обновления, источник набора и единицы измерения записывают как метаданные. Они описывают весь набор или его поля. Цена конкретной покупки и температура отдельного наблюдения относятся к значениям в строках.


# Чтение CSV

CSV — это текстовая таблица. При чтении CSV нужно согласовать разделитель и кодировку с файлом. Ошибочный разделитель объединяет несколько столбцов, а неверная кодировка искажает буквы. В таблице могут остаться пропуски. Даты проверяют отдельно.


# API веб-сервиса

API — это интерфейс для обмена запросами и ответами между программами. Программа может получить данные о маршруте через API веб-сервиса и сохранить результат как CSV.


# Поездки в метро

В примере с поездками метро поле publisher указывает издателя всего набора. В каждой строке city обозначает город наблюдения, year — год, metro_daily_riders_mln — среднее число пассажиров в день в миллионах.


# Подсчёт пропусков в pandas

Вызов df.isna().sum() считает пропуски по столбцам таблицы. df.head() показывает первые строки, df.dtypes возвращает типы столбцов, а pd.merge() объединяет таблицы.


# Полуструктурированные форматы

В этом примере к полуструктурированным данным относятся JSON, XML и HTML. Ключи и теги помогают найти значения. CSV здесь служит примером табличного формата.
